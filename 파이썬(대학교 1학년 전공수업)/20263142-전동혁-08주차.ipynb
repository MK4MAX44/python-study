```python
dinner_foods = []

while "먹고싶은거 생각중!~":
    food = input("저녁에 먹고 싶은 음식 입력 (그만 입력하면 종료)>> ")
    if food == "그만":
        break
    dinner_foods.append(food)

print(f"저녁에 먹고 싶은 음식 목록: {dinner_foods}")
```

```python
# 위 코드는 "그만"을 입력할 때까지 무한히 반복(break로만 탈출)
# 이번엔 조건식이 있는 while문으로 "최대 5개까지만" 입력받도록 제한

dinner_foods = []
count = 0

while count < 5:
    food = input(f"저녁에 먹고 싶은 음식 입력 ({count + 1}/5, 그만 입력하면 종료)>> ")
    if food == "그만":
        break
    dinner_foods.append(food)
    count += 1

print(f"저녁에 먹고 싶은 음식 목록 ({len(dinner_foods)}개): {dinner_foods}")
```
